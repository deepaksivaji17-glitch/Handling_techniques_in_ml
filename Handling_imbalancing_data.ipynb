# 1-import necessary data

```python
from sklearn import datasets
import pandas as pd
```

# 2- import dataset

```python
breast_cancer_data=datasets.load_breast_cancer()
breast_cancer_data
```

```python
breast_cancer_data.target
```

```python
breast_cancer_data_df=pd.DataFrame(data=breast_cancer_data.data,columns=breast_cancer_data.feature_names)
breast_cancer_data_df["target"]=breast_cancer_data.target
breast_cancer_data_df
```

```python
print(breast_cancer_data.DESCR)
```

# 3 data understanding

```python
breast_cancer_data_df.shape
```

```python
breast_cancer_data_df.isna().sum()
```

```python
breast_cancer_data_df.dtypes
```

# 4 model building

```python
X=breast_cancer_data_df.drop(labels="target",axis=1)
y=breast_cancer_data_df["target"]
```

```python
X.shape,y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.02,random_state=1234,shuffle=True,stratify=y)
```

```python
print("Training sample:",X_train,y_train.shape)
print("Test sample:",X_test,y_test.shape)
```

```python
y.value_counts()
```

```python
from matplotlib import pyplot as plt
plt.pie(x=y.value_counts(),labels=["Benign cancer","Malignant cancer"],autopct="%1.0f%%")
plt.show()
```

```python
plt.pie(x=y_train.value_counts(),labels=["Benign cancer","Malignant cancer"],autopct="%1.0f%%")
plt.show()
```

```python
plt.pie(x=y_test.value_counts(),labels=["Benign cancer","Malignant cancer"],autopct="%1.0f%%")
plt.show()
```

# 5 model training

## training data

```python
%%time
from sklearn.tree import DecisionTreeClassifier
dt_model=DecisionTreeClassifier( criterion='gini',max_depth=3,class_weight={0:2,1:1})
dt_model
```

```python
dt_model.fit(X_train,y_train)
```

# 6 model testing || 7.model evaluation
## training data

```python
y_pred_train=dt_model.predict(X_train)
y_pred_train
```

```python
 from sklearn.metrics import confusion_matrix,classification_report
```

```python
print(classification_report(y_train,y_pred_train))
```

```python
print(confusion_matrix(y_train,y_pred_train))
```

# test data

```python
y_pred_test=dt_model.predict(X_test)
y_pred_test
```

```python
print(classification_report(y_test,y_pred_test))
```

```python
print(confusion_matrix(y_test,y_pred_test))
```
